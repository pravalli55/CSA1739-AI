```python
import random

random.seed(42)

POPULATION_SIZE = 10
GENERATIONS = 30
MUTATION_RATE = 0.1

def create_individual():
    return [random.randint(0, 1) for _ in range(5)]

def decode(individual):
    return int("".join(map(str, individual)), 2)

def fitness(individual):
    x = decode(individual)
    return x ** 2

population = [
    create_individual()
    for _ in range(POPULATION_SIZE)
]

for generation in range(GENERATIONS):
    new_population = []

    best = max(population, key=fitness)
    new_population.append(best[:])

    while len(new_population) < POPULATION_SIZE:
        parent1 = max(random.sample(population, 3), key=fitness)
        parent2 = max(random.sample(population, 3), key=fitness)

        point = random.randint(1, 4)
        child = parent1[:point] + parent2[point:]

        for i in range(5):
            if random.random() < MUTATION_RATE:
                child[i] = 1 - child[i]

        new_population.append(child)

    population = new_population

best = max(population, key=fitness)

print("Best x:", decode(best))
print("Maximum Fitness:", fitness(best))
```
