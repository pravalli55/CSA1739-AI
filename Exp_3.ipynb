```python
from collections import deque

def water_jug(a, b, target):
    q = deque([(0, 0, [])])
    visited = set()

    while q:
        x, y, path = q.popleft()

        if (x, y) in visited:
            continue
        visited.add((x, y))
        path = path + [(x, y)]

        if x == target or y == target:
            print("Steps:")
            for step in path:
                print(step)
            return

        q.extend([
            (a, y, path), (x, b, path),
            (0, y, path), (x, 0, path),
            (max(0, x-(b-y)), min(b, x+y), path),
            (min(a, x+y), max(0, y-(a-x)), path)
        ])

    print("No solution")

water_jug(4, 3, 2)
```
