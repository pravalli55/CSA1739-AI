```python
from collections import deque

start = (3, 3, 1)
goal = (0, 0, 0)
moves = [(1, 0), (2, 0), (0, 1), (0, 2), (1, 1)]

q = deque([(start, [start])])
visited = {start}

while q:
    (m, c, boat), path = q.popleft()

    if (m, c, boat) == goal:
        for state in path:
            print(state)
        break

    for dm, dc in moves:
        if boat == 1:
            nm, nc, nb = m-dm, c-dc, 0
        else:
            nm, nc, nb = m+dm, c+dc, 1

        if 0 <= nm <= 3 and 0 <= nc <= 3:
            if (nm == 0 or nm >= nc) and (nm == 3 or 3-nm >= 3-nc):
                s = (nm, nc, nb)
                if s not in visited:
                    visited.add(s)
                    q.append((s, path + [s]))
else:
    print("No solution")
```
