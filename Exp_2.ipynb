```python
def solve(row, cols, pos_diag, neg_diag, board):
    if row == 8:
        for r in board:
            print(". " * r + "Q " + ". " * (7-r))
        return True

    for col in range(8):
        if col not in cols and row-col not in pos_diag and row+col not in neg_diag:
            if solve(row+1, cols | {col},
                     pos_diag | {row-col},
                     neg_diag | {row+col}, board + [col]):
                return True
    return False

solve(0, set(), set(), set(), [])
```
