```python
board = [' '] * 9

def display():
    for i in range(0, 9, 3):
        print(board[i], '|', board[i+1], '|', board[i+2])
        if i < 6:
            print('--+---+--')

def winner(p):
    lines = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]
    return any(all(board[i] == p for i in line)
               for line in lines)

for turn in range(9):
    display()
    p = 'X' if turn % 2 == 0 else 'O'

    while True:
        try:
            pos = int(input(f"Player {p}, enter position (1-9): ")) - 1
            if 0 <= pos < 9 and board[pos] == ' ':
                break
            print("Invalid position. Try again.")
        except ValueError:
            print("Enter a number from 1 to 9.")

    board[pos] = p

    if winner(p):
        display()
        print("Player", p, "wins!")
        break
else:
    display()
    print("Game Draw!")
```
