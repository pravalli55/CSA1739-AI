```python
def minimax(board, is_max):
    if 'X' in board and board.count('X') >= 3:
        if win(board, 'X'):
            return 1
    if win(board, 'O'):
        return -1
    if ' ' not in board:
        return 0

    if is_max:
        best = -float('inf')
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'X'
                best = max(best, minimax(board, False))
                board[i] = ' '
        return best
    else:
        best = float('inf')
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'O'
                best = min(best, minimax(board, True))
                board[i] = ' '
        return best

def win(b, p):
    lines = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]
    return any(all(b[i] == p for i in line) for line in lines)

board = [' '] * 9
best_score = -float('inf')
best_move = None

for i in range(9):
    board[i] = 'X'
    score = minimax(board, False)
    board[i] = ' '

    if score > best_score:
        best_score = score
        best_move = i

print("Best move for X:", best_move + 1)
```
