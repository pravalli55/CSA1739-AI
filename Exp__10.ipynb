```python
import heapq

graph = {
    'A': [('B', 1), ('C', 3)],
    'B': [('D', 1), ('E', 5)],
    'C': [('E', 2)],
    'D': [('E', 1)],
    'E': []
}

h = {'A': 4, 'B': 3, 'C': 2, 'D': 1, 'E': 0}

def astar(start, goal):
    q = [(h[start], 0, start, [start])]

    while q:
        f, g, node, path = heapq.heappop(q)

        if node == goal:
            print("Path:", path)
            print("Cost:", g)
            return

        for child, cost in graph[node]:
            new_g = g + cost
            heapq.heappush(
                q, (new_g + h[child], new_g,
                    child, path + [child])
            )

astar('A', 'E')
```
