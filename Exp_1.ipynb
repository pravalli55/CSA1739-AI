```python
from collections import deque

start = (1, 2, 3, 4, 0, 6, 7, 5, 8)
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

q = deque([(start, [])])
visited = {start}

while q:
    state, path = q.popleft()

    if state == goal:
        print("Solution:", path + [state])
        break

    z = state.index(0)
    for n in [z-3, z+3, z-1, z+1]:
        if 0 <= n < 9 and not (abs(n-z) == 1 and z//3 != n//3):
            s = list(state)
            s[z], s[n] = s[n], s[z]
            s = tuple(s)
            if s not in visited:
                visited.add(s)
                q.append((s, path + [state]))
else:
    print("No solution")
```
