```python
room = {'A': 'Dirty', 'B': 'Dirty'}
position = 'A'

while 'Dirty' in room.values():
    if room[position] == 'Dirty':
        room[position] = 'Clean'
        print("Cleaning room", position)
    elif position == 'A':
        position = 'B'
        print("Moving to room B")
    else:
        position = 'A'
        print("Moving to room A")

print("All rooms are clean")
```
