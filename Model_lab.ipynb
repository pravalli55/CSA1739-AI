```python

def dls(graph, node, goal, limit):
    if node == goal:
        return [node]

    if limit <= 0:
        return None

    for child in graph.get(node, []):
        result = dls(graph, child, goal, limit - 1)
        if result:
            return [node] + result

    return None


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

start = 'A'
goal = 'G'
limit = 2

path = dls(graph, start, goal, limit)

if path:
    print("Goal found!")
    print("Path:", " -> ".join(path))
else:
    print("Goal not found within depth limit.")
```

```python

import heapq

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def heuristic(state):
    distance = 0

    for i, tile in enumerate(state):
        if tile != 0:
            target = goal.index(tile)
            distance += abs(i // 3 - target // 3)
            distance += abs(i % 3 - target % 3)

    return distance


def get_neighbors(state):
    zero = state.index(0)
    row, col = divmod(zero, 3)
    moves = []

    for dr, dc in [(1, 0), (-1, 0),
                   (0, 1), (0, -1)]:
        nr, nc = row + dr, col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_pos = nr * 3 + nc
            new_state = list(state)
            new_state[zero], new_state[new_pos] = (
                new_state[new_pos], new_state[zero]
            )
            moves.append(tuple(new_state))

    return moves


def astar(start):
    queue = [(heuristic(start), 0, start, [start])]
    best_cost = {start: 0}

    while queue:
        f, g, state, path = heapq.heappop(queue)

        if state == goal:
            return path

        if g != best_cost.get(state):
            continue

        for neighbor in get_neighbors(state):
            new_g = g + 1

            if new_g < best_cost.get(neighbor, float('inf')):
                best_cost[neighbor] = new_g
                new_f = new_g + heuristic(neighbor)

                heapq.heappush(
                    queue,
                    (new_f, new_g, neighbor, path + [neighbor])
                )

    return None


start = (1, 2, 3,
         4, 5, 6,
         0, 7, 8)

path = astar(start)

if path:
    for step, state in enumerate(path):
        print("Step", step)
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
    print("Total moves:", len(path) - 1)
else:
    print("No solution found.")
```

```python

import numpy as np
import random

Q = np.zeros((4, 4))

alpha = 0.8
gamma = 0.9
epsilon = 0.2
episodes = 1000
goal = 3

# Actions: 0=Up, 1=Down, 2=Left, 3=Right

def next_state(state, action):
    row, col = divmod(state, 2)

    if action == 0 and row > 0:
        row -= 1
    elif action == 1 and row < 1:
        row += 1
    elif action == 2 and col > 0:
        col -= 1
    elif action == 3 and col < 1:
        col += 1

    return row * 2 + col


for episode in range(episodes):
    state = 0

    while state != goal:
        if random.random() < epsilon:
            action = random.randint(0, 3)
        else:
            action = np.argmax(Q[state])

        new_state = next_state(state, action)
        reward = 10 if new_state == goal else -1

        Q[state, action] += alpha * (
            reward + gamma * np.max(Q[new_state])
            - Q[state, action]
        )

        state = new_state

print("Q-Table:")
print(np.round(Q, 2))

# Display learned path
state = 0
path = [state]

for _ in range(10):
    if state == goal:
        break

    action = np.argmax(Q[state])
    state = next_state(state, action)
    path.append(state)

print("Learned Path:", " -> ".join(map(str, path)))
```
