```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

ratings = pd.DataFrame(
    {
        "Movie_A": [5, 4, 0, 0],
        "Movie_B": [4, 5, 0, 0],
        "Movie_C": [0, 0, 5, 4],
        "Movie_D": [0, 0, 4, 5]
    },
    index=["User1", "User2", "User3", "User4"]
)

print("User Rating Matrix:")
print(ratings)

similarity = cosine_similarity(ratings)

similarity_df = pd.DataFrame(
    similarity,
    index=ratings.index,
    columns=ratings.index
)

target_user = "User1"
user_similarity = similarity_df.loc[target_user].drop(target_user)

unseen_items = ratings.columns[ratings.loc[target_user] == 0]
recommendations = {}

for item in unseen_items:
    numerator = 0
    denominator = 0

    for user in ratings.index:
        if user == target_user:
            continue

        rating = ratings.loc[user, item]
        sim = user_similarity[user]

        if rating > 0:
            numerator += sim * rating
            denominator += abs(sim)

    if denominator > 0:
        recommendations[item] = numerator / denominator

print("\nRecommendations for", target_user)

for item, score in sorted(
    recommendations.items(),
    key=lambda x: x[1],
    reverse=True
):
    print(item, "Predicted Rating:", round(score, 2))
```
